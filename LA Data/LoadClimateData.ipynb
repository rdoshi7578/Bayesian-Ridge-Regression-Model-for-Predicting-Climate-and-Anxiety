```python
import pandas as pd

# 1. Load CSV
file_path = 'C:/Users/Jack/Downloads/3B_LosAngeles_MY_2001-2020.csv'
df = pd.read_csv(file_path)

# 2. Clean and parse datetime
# Replace 'time_lst' with your actual datetime column name if different
datetime_col = 'time_lst'
df[datetime_col] = df[datetime_col].str.strip()  # remove extra spaces
df[datetime_col] = pd.to_datetime(df[datetime_col], errors='coerce')  # convert to datetime

# Warn about any rows that failed conversion
failed_rows = df[df[datetime_col].isna()]
if not failed_rows.empty:
    print(f"Warning: {len(failed_rows)} rows could not be parsed as datetime.")

# 3. Set datetime as index
df.set_index(datetime_col, inplace=True)

# 4. Compute monthly averages
monthly_avg = df.resample('ME').mean()

# 5. Preview results
print(monthly_avg.head())

# Optional: save monthly averages to a new CSV
# monthly_avg.to_csv('C:/Users/Jack/Downloads/3B_LosAngeles_monthly_avg.csv')
```
