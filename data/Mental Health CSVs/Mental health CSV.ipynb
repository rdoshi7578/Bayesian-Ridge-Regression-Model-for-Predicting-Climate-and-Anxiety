```python
import pandas as pd

# 1. Load the dataset
# Replace 'cdc_data.csv' with your actual filename
df = pd.read_csv('mental_health_ED.csv')

# 2. Create the Age Data CSV
# Filters for rows where demographics_type is 'Age'
age_df = df[df['demographics_type'] == 'Age']
age_df.to_csv('mental_health_by_age.csv', index=False)

# 3. Create the Ethnicity Data CSV
# Filters for rows where demographics_type is 'RaceEthnicity'
ethnicity_df = df[df['demographics_type'] == 'RaceEthnicity']
ethnicity_df.to_csv('mental_health_by_ethnicity.csv', index=False)

# 4. Create the Sex Data CSV
# Filters for rows where demographics_type is 'Sex'
sex_df = df[df['demographics_type'] == 'Sex']
sex_df.to_csv('mental_health_by_sex.csv', index=False)

print("Files created successfully:")
print("- mental_health_by_age.csv")
print("- mental_health_by_ethnicity.csv")
print("- mental_health_by_sex.csv")
```
